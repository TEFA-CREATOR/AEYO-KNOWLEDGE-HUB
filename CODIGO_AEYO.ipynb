**USUARIOS**
---



```
import re
import os

class clsUsuarios:
    """Esta es la parte que se encargara de registrar y validar usuarios."""

    def __init__(self, nombre, apellido, documento, correo, tiempo_prestamo):
        self.nombre = nombre
        self.apellido = apellido
        self.documento = documento
        self.correo = correo
        self.tiempo_prestamo = tiempo_prestamo

    def validar_nombre(self):
        """Valida que el nombre tenga al menos 3 letras y no contenga números."""
        return self.nombre.isalpha() and len(self.nombre) >= 3

    def validar_apellido(self):
        """Valida que el apellido tenga al menos 3 letras y no contenga números."""
        return self.apellido.isalpha() and len(self.apellido) >= 3

    def validar_documento(self):
        """Verifica que el documento solo tenga números y una longitud válida."""
        return self.documento.isdigit() and 3 <= len(self.documento) <= 15

    def validar_correo(self):
        """Verifica que el correo tenga formato válido terminado en .com"""
        patron = r'^[\w\.-]+@[\w\.-]+\.com$'
        return re.match(patron, self.correo)

    def validar_tiempo(self):
        """Verifica que el tiempo de préstamo sea permitido."""
        return self.tiempo_prestamo in [5, 10, 15, 30]

    def guardar_usuario(self):
        """Guarda el usuario en un archivo plano si todas las validaciones son correctas."""
        if (self.validar_nombre() and self.validar_apellido() and
            self.validar_documento() and self.validar_correo() and
            self.validar_tiempo()):

            # Ensure the directory exists
            os.makedirs("src/datos", exist_ok=True)

            with open("src/datos/usuarios.txt", "a", encoding="utf-8") as archivo:
                archivo.write(f"{self.nombre},{self.apellido},{self.documento},{self.correo},{self.tiempo_prestamo}\n")

            print("Usuario registrado correctamente")
        else:
            print("Error en los datos ingresados")
```

```
import random
import os

class Item:
    """Este punto se encargara de gestionar los ítems que se prestan, usando un listado de categorias que nos ayudaran a clasificar nuestro software.
    Se usaron diccionarios para mejorar la organización y acceso eficiente a los datos en categorías y estados"""

    categorias = {

        "Revistas": {
            "codigo": "REV",
            "descripcion": "Revistas de todo tipo"
        },

        "Libros": {
            "codigo": "LIB",
            "descripcion": "Libros y textos"
        },

        "Infantiles": {
            "codigo": "INF",
            "descripcion": "CD, DVD y multimedia"
        },

        "Tesis": {
            "codigo": "TES",
            "descripcion": "Trabajos de tesis de pregrado y posgrado"
        },

        "Periodicos": {
            "codigo": "PER",
            "descripcion": "Periodicos y publicaciones"
        },

        "Miscelaneo": {
            "codigo": "MIS",
            "descripcion": "Objetos varios"
        }
    }
    estados = {
        "Nuevo": {
            "codigo": "N",
            "descripcion": "Nuevo"
        },

        "Usado": {
            "codigo": "U",
            "descripcion": "Usado"
       },

        "Deteriorado": {
            "codigo": "D",
            "descripcion": "Deteriorado"
        },

        "Otro": {
            "codigo": "O",
            "descripcion": "Otro"
        }
    }

    def __init__(self, nombre, categoria, precio, nivel_estado):
        self.nombre = nombre
        self.categoria = categoria
        self.precio = precio
        self.estado = nivel_estado
        self.id_item = self.generar_id()
    def generar_id(self):
        """Genera un identificador único usando la categoría."""
        prefijo = self.categorias[self.categoria]["codigo"]
        numero = random.randint(1000, 9999)
        return f"{prefijo}{numero}"
    def guardar_item(self):
        """ Y aca guarda el ítem en archivo plano"""
        # Ensure the directory exists
        os.makedirs("src/datos", exist_ok=True)
        with open("src/datos/items.txt", "a", encoding="utf-8") as archivo:
            archivo.write(
                f"{self.id_item},"
                f"{self.nombre},"
                f"{self.categoria},"
                f"{self.precio},"
                f"{self.estado}\n"
            )
        print("Item registrado correctamente")
```

**PRESTAMOS**
---



```
from datetime import datetime, timedelta
import os

class clsPrestamo:
    """lo que buscamos es registramos un préstamo en el sistema pero debemos asegurarnos
    que sea a un usuario registrado."""

    def __init__(self, documento_usuario, id_item):
        self.documento_usuario = documento_usuario
        self.id_item = id_item
        self.fecha_prestamo = datetime.now() - timedelta(hours=5)

    def registrar_prestamo(self):
        # Ensure the directory exists for all file operations
        os.makedirs("src/datos", exist_ok=True)

        with open("src/datos/usuarios.txt", "r", encoding="utf-8") as archivo:
          usuarios = archivo.readlines()
        usuario_existe = False
        for linea in usuarios:
          datos_usuario = linea.strip().split(",")
          documento_usuario = datos_usuario[2]
          if documento_usuario == self.documento_usuario:
            usuario_existe = True
            break
        if not usuario_existe:
          print("USUARIO NO REGISTRADO")
          return

        with open("src/datos/prestamos.txt", "r", encoding="utf-8") as archivo:
            prestamos = archivo.readlines()
        for linea in prestamos:
            datos = linea.strip().split(",")
            item_prestado = datos[1]
            estado = datos[3]
            if item_prestado == self.id_item and estado == "ACTIVO":
              print("ESTE ITEM YA ESTA PRESTADO")
              return
        with open("src/datos/prestamos.txt", "a", encoding="utf-8") as archivo:
          archivo.write(
            f"{self.documento_usuario},"
            f"{self.id_item},"
            f"{self.fecha_prestamo},"
            f"ACTIVO\n"
        )
        print("PRESTAMO REGISTRADO")
```

**DEVOLUCIONES**
---



```
from datetime import datetime, timedelta
import os

"""Vericamos que el usuario no tenga devoluciones y de ser asi genera el certificado de
devolución del ítem, si no tiene prestamos lanza un mensaje y no un certificado"""

def registrar_devolucion(self):
    # Ensure the directory exists for all file operations
    os.makedirs("src/datos", exist_ok=True)

    with open("src/datos/prestamos.txt", "r", encoding="utf-8") as archivo:
        prestamos = archivo.readlines()
    prestamo_encontrado = False
    nuevas_lineas = []
    for linea in prestamos:
        datos = linea.strip().split(",")
        documento = datos[0]
        item = datos[1]
        fecha = datos[2]
        estado = "ACTIVO"
        if len(datos) > 3:
            estado = datos[3]

        if (
            documento == self.documento_usuario
            and item == self.id_item
            and estado == "ACTIVO"
        ):
            prestamo_encontrado = True
            nuevas_lineas.append(
                f"{documento},{item},{fecha},DEVUELTO\n"
            )
        else:

            if len(datos) < 4:
                nuevas_lineas.append(f"{documento},{item},{fecha},ACTIVO\n")
            else:
                nuevas_lineas.append(linea)

    with open("src/datos/prestamos.txt", "w", encoding="utf-8") as archivo:
        archivo.writelines(nuevas_lineas)
    if prestamo_encontrado:
        certificado = f"Certificado_{self.documento_usuario}_{self.id_item}.txt"
        with open(certificado, "w", encoding="utf-8") as archivo:
            archivo.write("CERTIFICADO DE DEVOLUCION\n")
            archivo.write(f"Usuario: {self.documento_usuario}\n")
            archivo.write(f"Item: {self.id_item}\n")
            archivo.write(f"Fecha devolucion: " f"{datetime.now()- timedelta(hours=5)}\n")

        print("Devolucion registrada")
    else:
        print("Usted no tiene prestamos activos para devolver")
```

**ADMINISTRADOR**
---



```
class Administrador:
    """Este será el espacio para el acceso administrativo."""

    usuarios_admin = {
        "Admin": {
            "clave": "1234",
            "rol": "Administrador principal"
        },
        "Recepcion": {
            "clave": "5678",
            "rol": "Gestor de prestamos"
        }
    }
    def login(self, usuario, clave):
        """Verifica el usuario y la contraseña."""
        if usuario in self.usuarios_admin:
            if self.usuarios_admin[usuario]["clave"] == clave:
                print("Acceso permitido")
                print("Rol:",
                      self.usuarios_admin[usuario]["rol"])
                return True
        print("Acceso denegado")
        return False
    def total_registros(self, archivo):
        """Cuenta registros de un archivo."""
        try:
            with open(archivo, "r", encoding="utf-8") as f:
                return len(f.readlines())
        except:
            return 0
```

**UTILIDADES, FACTURAS Y ESTADISTICAS GENERALES**

```
import csv
import os
""" La idea de este codigo es saber la frecuencia de los prestamos, devoluciones y ventas, que nos muestre las estadisticas y ademas de exportar TXT a CSV."""
estadisticas = {

    "prestamos": 0,
    "devoluciones": 0,
    "ventas": 0,
    "total_pagado": 0
}
def exportar_csv(origen, destino):
    # Ensure the directory exists for 'src/datos' if it's part of the path
    # And also for the destination directory if it's not the current one.
    output_dir = os.path.dirname(destino)
    if output_dir and not os.path.exists(output_dir):
        os.makedirs(output_dir, exist_ok=True)

    with open(origen, "r", encoding="utf-8") as archivo:

        lineas = archivo.readlines()

    with open(destino, "w", newline="", encoding="utf-8") as csvfile:

        writer = csv.writer(csvfile)

        for linea in lineas:

            writer.writerow(linea.strip().split(","))

    print("Archivo exportado correctamente")


def mostrar_estadisticas():

    print("\n===== ESTADISTICAS =====")

    print("Prestamos:",
          estadisticas["prestamos"])

    print("Devoluciones:",
          estadisticas["devoluciones"])

    print("Ventas:",
          estadisticas["ventas"])

    print("Total pagado:",
          estadisticas["total_pagado"])
```

```
from datetime import datetime, timedelta
"""Genera automáticamente una factura por pasarse del tiempo del préstamo."""

def generar_factura(nombre_usuario, item, precio):
    impuesto = precio * 0.19
    total = precio + impuesto

    nombre_archivo = f"Factura_{nombre_usuario}_{item}.txt"

    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        archivo.write("FACTURA DE VENTA\n")
        archivo.write("Motivo: Prestamo superior a 30 dias\n")
        archivo.write(f"Usuario: {nombre_usuario}\n")
        archivo.write(f"Item: {item}\n")
        archivo.write(f"Subtotal: ${precio: .2f}\n")
        archivo.write(f"Impuesto IVA 19%: ${impuesto:.2f}\n")
        archivo.write(f"Total: ${total:.2f}\n")
        archivo.write(f"Fecha: " f"{datetime.now() - timedelta(hours=5)}\n")

    print("FACTURA GENERADA")
```

```
def menu():
    while True:
        print("\n===== AEYO KNOWLEDGE HUB =====")
        print("1. Registrar usuario")
        print("2. Registrar item")
        print("3. Registrar prestamo")
        print("4. Registrar devolucion")
        print("5. Exportar CSV")
        print("6. Administrador")
        print("7. Salir")

        opcion = input("Seleccione una opcion: ")

        if opcion == "1":
            while True:
                nombre = input("Nombre: ")
                apellido = input("Apellido: ")
                documento = input("Documento: ")
                correo = input("Correo: ")
                tiempo = int(
                    input("Tiempo de prestamo (5,10,15,30): ")
                )
                print("\n===== CONFIRMAR DATOS =====")
                print(f"Nombre: {nombre}")
                print(f"Apellido: {apellido}")
                print(f"Documento: {documento}")
                print(f"Correo: {correo}")
                print(f"Tiempo de prestamo: {tiempo}")
                confirmar = input(
                    "\n¿Los datos son correctos? (S/N): "
                ).upper()
                if confirmar == "S":
                    usuario = clsUsuarios(
                        nombre,
                        apellido,
                        documento,
                        correo,
                        tiempo
                    )
                    usuario.guardar_usuario()
                    break
                else:
                    print("\nIngrese nuevamente los datos\n")
        elif opcion == "2":
            while True:
                nombre_item = input("Nombre del item: ")
                print("\nCategorias disponibles:")
                for nombre_categoria, datos in Item.categorias.items():
                    print(f"{datos['codigo']} - {nombre_categoria}")
                codigo_categoria = input(
                    "Ingrese codigo de categoria: "
                ).upper()
                categoria_encontrada = None
                for nombre_categoria, datos in Item.categorias.items():
                    if datos["codigo"] == codigo_categoria:
                        categoria_encontrada = nombre_categoria
                        break
                precio = float(input("Precio: "))
                print("\nEstados disponibles:")
                for nombre_estado, datos in Item.estados.items():
                    print(f"{datos['codigo']} - {nombre_estado}")
                codigo_estado = input(
                    "Ingrese codigo de estado: "
                ).upper()
                estado_encontrado = None
                for nombre_estado, datos in Item.estados.items():
                    if datos["codigo"] == codigo_estado:
                        estado_encontrado = nombre_estado
                        break
                print("\n===== CONFIRMAR DATOS =====")
                print(f"Nombre: {nombre_item}")
                print(f"Categoria: {categoria_encontrada}")
                print(f"Precio: {precio}")
                print(f"Estado: {estado_encontrado}")
                confirmar = input(
                    "\n¿Los datos son correctos? (S/N): "
                ).upper()
                if confirmar == "S":
                    item = Item(
                        nombre_item,
                        categoria_encontrada,
                        precio,
                        estado_encontrado
                    )
                    item.guardar_item()
                    break
                else:
                    print("\nIngrese nuevamente los datos\n")

        elif opcion == "3":
                documento = input("Documento usuario: ")
                id_item = input("ID item: ")
                prestamo = clsPrestamo(documento, id_item)
                prestamo.registrar_prestamo()


        elif opcion == "4":
                documento = input("Documento usuario: ")
                id_item = input("ID item: ")
                devolucion = clsPrestamo(documento, id_item)
                registrar_devolucion(devolucion)

        elif opcion == "5":
                print("\n===== EXPORTAR CSV =====")
                print("1. Exportar usuarios")
                print("2. Exportar items")
                print("3. Exportar prestamos")
                opcion_csv = input("Seleccione una opcion: ")
                if opcion_csv == "1":
                    exportar_csv(
                        "src/datos/usuarios.txt",
                        "usuarios.csv")
                    print("USUARIOS EXPORTADOS CORRECTAMENTE")
                elif opcion_csv == "2":
                    exportar_csv(
                        "src/datos/items.txt",
                        "items.csv")
                    print("ITEMS EXPORTADOS CORRECTAMENTE")
                elif opcion_csv == "3":
                    exportar_csv(
                        "src/datos/prestamos.txt",
                        "prestamos.csv")
                    print("PRESTAMOS EXPORTADOS CORRECTAMENTE")
                else:
                    print("Opcion invalida")
        elif opcion == "6":
                admin = Administrador()
                user = input("Usuario admin: ")
                clave = input("Clave: ")
                if admin.login(user, clave):
                    print("Acceso concedido")
                    print("Total usuarios:", admin.total_registros("src/datos/usuarios.txt"))
                    print("Total prestamos:", admin.total_registros("src/datos/prestamos.txt"))
                else:
                    print("Acceso denegado")
        elif opcion == "7":
                print("Saliendo...")
                break
        else:
                print("Opcion invalida")

menu()
```
